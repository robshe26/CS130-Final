```python
import csv

empty_list = []

total_city_mpg = 0
total_cars = 0

with open('cars (1).csv', 'r', encoding='utf-8') as file:
    reader = csv.reader(file)
    next(reader)  
    
    for row in reader:
        empty_list.append(row)

    for row in empty_list:
        city_mpg_str = row[8]
        
        city_mpg = float(city_mpg_str)

        total_city_mpg += city_mpg
        total_cars += 1

    average_city_mpg = total_city_mpg / total_cars

    print("Average City MPG:", round(average_city_mpg, 2))
```

```python
import pandas as pd 
cars_info = pd.read_csv('cars (1).csv')
cars_info.head()

```

```python
cars_info['Fuel Information.City mpg']
cars_info.iloc[[299, 305], 4]


#cars_info.loc(cars_info['Engine Information.Driveline'] == "Front-wheel drive") & (cars_info['Fuel Information.City mpg'] == 30) 


cars_info['City Mpg / Highway Mpg'] = cars_info['Fuel Information.City mpg'] / cars_info['Fuel Information.Highway mpg']

cars_info.head()

cars_info['Dimensions.Height'].describe()

cars_info['Fuel Information.City mpg'].value_counts()

def count_city_mpg(city_mpg_value):
    # Filter the DataFrame to count the number of cars with the specified City MPG value
    count = cars_info[cars_info['Fuel Information.City mpg'] == city_mpg_value].shape[0] # Get the number of rows that match the condition
    return count    

city_mpg_value = 30
count = count_city_mpg(city_mpg_value)
print(f"Number of cars with City MPG of {city_mpg_value}: {count}") 


```
