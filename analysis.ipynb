```python
import csv

empty_list = []

total_city_mpg = 0
total_cars = 0

with open('cars (1).csv', 'r', encoding='utf-8') as file:
    reader = csv.reader(file)
    next(reader)  
    
    for row in reader:
        empty_list.append(row)

    for row in empty_list:
        city_mpg_str = row[8]
        
        city_mpg = float(city_mpg_str)

        total_city_mpg += city_mpg
        total_cars += 1

    average_city_mpg = total_city_mpg / total_cars

    print("Average City MPG:", round(average_city_mpg, 2))
```

```python
cars_info.loc[cars_info['Highway_mpg'] > 100]
cars_clean = cars_info.drop(cars_info[cars_info['Highway_mpg'] > 100].index)
cars_clean['Highway_mpg'].describe()
```

```python
cars_clean['Volume'] = ((cars_clean['Length'] * cars_clean['Width'] * cars_clean['Height']) // 1728)
cars_clean.head()
```

```python
cars_clean['Volume'].describe()
```

```python
cars_clean['MPG_Gap'] = cars_clean['Highway_mpg'] - cars_clean['City_mpg']
cars_clean.head()
```

```python
cars_clean['MPG_Gap'].describe()
```

```python
# uses a dictionary to replace the long driveline descriptions with shorter abbreviations for easier analysis and visualization
cars_clean['Driveline'] = cars_clean['Driveline'].replace({
    'Front-wheel drive': 'FWD',
    'Rear-wheel drive': 'RWD',
    'All-wheel drive': 'AWD',
    'Four-wheel drive': '4WD'
})
cars_clean.head()
```

```python
cars_clean['avg_mpg'] = (cars_clean['City_mpg'] + cars_clean['Highway_mpg']) / 2
#used the goupby and map functions to calculate the average MPG for each driveline type and add it as a new column to the DataFrame
Driveline_avg = cars_clean.groupby('Driveline')['avg_mpg'].mean()
cars_clean['Driveline_avg'] = cars_clean['Driveline'].map(Driveline_avg)
cars_clean.head()
```

```python
cars_clean['Driveline_avg'].value_counts()
```

```python
# the groupby and map functions to calculate the average MPG gap for each driveline type and add it as a new column to the DataFrame
avg_gap_driveline = cars_clean.groupby('Driveline')['MPG_Gap'].mean()
cars_clean['avg_gap_driveline'] = cars_clean['Driveline'].map(avg_gap_driveline)
cars_clean.head()
```

```python
cars_clean['avg_gap_driveline'].value_counts()
```

```python
Driveline_avg.plot(
    x='Driveline',
    kind ='bar',
    figsize=(10, 6), 
    color=['red']

)    
plt.title('AVG Fuel Economy by Driveline')
plt.ylabel('AVG MPG')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('AVG_Fuel_Economy_by_Driveline.png')
plt.show()



```

```python
avg_gap_driveline.plot(
    x='Driveline',
    y='MPG_Gap',
    kind='bar',
    figsize=(10, 6),
    color=['green']
)
plt.title('How Extra Wheel Drive Affects the MPG Gap')
plt.ylabel('MPG Gap (Extra MPG on the Highway)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('driveline_bar.png')
plt.show()
```
